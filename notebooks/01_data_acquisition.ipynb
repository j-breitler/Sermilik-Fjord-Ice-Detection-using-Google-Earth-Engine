# WP1: Study Area & Data Acquisition

Set up the GEE session, load the AOI, and define Sentinel-1, Sentinel-2, and ERA5 collections for 2019–2024.

```python
import ee
import geemap
import sys
sys.path.insert(0, '..')
from src.utils import load_aoi, get_gee_project, date_range

ee.Initialize(project=get_gee_project())
```

## 1.1 Load AOI

```python
# Replace with your shapefile name
aoi = load_aoi()
print(aoi.getInfo())
```

## 1.2 Sentinel-2 Collection

```python
START = '2019-01-01'
END   = '2024-12-31'

s2 = (
    ee.ImageCollection('COPERNICUS/S2_SR_HARMONIZED')
    .filterBounds(aoi)
    .filterDate(START, END)
    .filter(ee.Filter.lt('CLOUDY_PIXEL_PERCENTAGE', 80))
)
print('S2 images:', s2.size().getInfo())
```

## 1.3 Sentinel-1 Collection

```python
s1 = (
    ee.ImageCollection('COPERNICUS/S1_GRD')
    .filterBounds(aoi)
    .filterDate(START, END)
    .filter(ee.Filter.listContains('transmitterReceiverPolarisation', 'VV'))
    .filter(ee.Filter.eq('instrumentMode', 'IW'))
    .select(['VV', 'VH'])
)
print('S1 images:', s1.size().getInfo())
```

## 1.4 ERA5 Collection

```python
era5 = (
    ee.ImageCollection('ECMWF/ERA5_LAND/DAILY_AGGR')
    .filterBounds(aoi)
    .filterDate(START, END)
    .select(['temperature_2m', 'u_component_of_wind_10m', 'v_component_of_wind_10m'])
)
print('ERA5 images:', era5.size().getInfo())
```
